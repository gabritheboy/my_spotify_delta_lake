```python

%sql
-- Top 20 songs played in the last seven days

CREATE or REPLACE  VIEW my_spotify.gold.top_20_played_last_7_days AS
SELECT 
p.track_id,
p.track_name,
a.name as artist_name,
t.popularity as popularity,
COUNT(*) as plays
FROM my_spotify.silver.user_recent_played p
JOIN my_spotify.silver.artists a
ON p.first_artist_id = a.id
JOIN my_spotify.silver.tracks t
ON p.track_id = t.id
WHERE p.played_at >= current_date() - INTERVAL 7 DAYS
GROUP BY p.track_id,
p.track_name,
a.name,
t.popularity
ORDER BY plays DESC
LIMIT 20;

```

```python
%sql

--TOP 5 artists by number of tracks played in the last seven days

CREATE or REPLACE VIEW my_spotify.gold.top_5_artists_last_7_days AS
SELECT 
artist_name,
SUM(plays) as total_plays_last_7_days
from my_spotify.gold.top_20_played_last_7_days
GROUP BY artist_name
```

```python
%sql
-- TOP 10 Artists of all time

CREATE OR REPLACE VIEW my_spotify.gold.top_10_artists_all_time AS
SELECT 
p.first_artist_id as artist_id,
a.name as artist_name,
a.popularity as artist_popularity,
COUNT(*) as total_plays
FROM my_spotify.silver.user_recent_played p
JOIN my_spotify.silver.artists a
ON p.first_artist_id = a.id
GROUP BY p.first_artist_id,
a.name,
a.popularity
ORDER BY total_plays DESC
LIMIT 10;
```
